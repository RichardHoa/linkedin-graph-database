# LinkedIn Graph RAG with Ollama

This notebook demonstrates the Graph RAG pipeline using **Ollama** for text-to-cypher generation. 
We use the `Gemma 3 4B Instruct` model via Ollama for efficient, quantized inference.

```python
import os
from dotenv import load_dotenv
import sys

# Add Project directory to path to import graph_rag
sys.path.append('../Project')

load_dotenv()

# Verify environment
if not os.getenv("NEO4J_SECRET"):
    print("Warning: NEO4J_SECRET not found in .env")
if not os.getenv("OPENAI_API_KEY"):
    print("Warning: OPENAI_API_KEY not found in .env (needed for embeddings if not using Ollama for them)")
```

```python
from graph_rag import GraphRAGPipeline

# Initialize the pipeline (Uses Ollama for completions)
pipeline = GraphRAGPipeline()
```

## Example Query

Run a natural language query to generate and execute Cypher.

```python
user_query = "Find all professionals who work at Google and have experience in Python."
result = pipeline.run(user_query)

print(f"Generated Cypher:\n{result['cypher_query']}\n")
print(f"Results:\n{result['final_data']}")
```
